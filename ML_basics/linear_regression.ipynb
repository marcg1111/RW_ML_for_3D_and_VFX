```python
import torch
import matplotlib.pyplot as plt
```

```python
data = {
    3.5:      18,
    3.69:     15,
    3.44:     18,
    3.43:     16.5,
    4.34:     15.5,
    4.42:     14.5,
    2.37:     24.5
    }
```

```python
weights = torch.tensor(list(data.keys()), dtype=torch.float32)
miles_per_gallon = torch.tensor(list(data.values()), dtype=torch.float32)

y_max = miles_per_gallon.max().item() + 1
print(y_max)
```

```python
plt.scatter(weights.numpy(), miles_per_gallon.numpy())
plt.xlabel("Pounds in 1000s")
plt.ylabel("Miles per gallon")
plt.ylim(0, 35)
plt.xlim(0, 5)
plt.title("Weight vs. Miles per gallon")
plt.show()
```

```python
import numpy as np

x = weights.numpy()
y = miles_per_gallon.numpy()

slope, intercept = np.polyfit(x, y, 1)
line_x = np.linspace(0 , 5, 100)
line_y = slope * line_x + intercept

plt.scatter(x, y)
plt.plot(line_x, line_y, color='red', label=f"y = {slope:.2f}x + {intercept:.2f}")
plt.xlabel("Pounds in 1000s")
plt.ylabel("Miles per gallon")
plt.ylim(0, 35)
plt.xlim(0, 5)
plt.title("Weight vs. Miles per gallon")
plt.show()
```

```python
def calculate_miles_per_gallons(weight):
    result = slope * weight + intercept
    print(f"slope: {slope}")
    print(f"intercept: {intercept}")
    print(f"result: {result}")

calculate_miles_per_gallons(4)
```
