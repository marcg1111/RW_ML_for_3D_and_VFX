```python
import jax.numpy as jnp
```

```python
def relu(x):
    return jnp.maximum(0, x)

def softmax(x):
    return jnp.exp(x) / jnp.sum(jnp.exp(x), axis=0)

def cross_entropy(x, y):
    return -jnp.sum(y * jnp.log(x))
```

```python
x = jnp.arange(6.0)
print(relu(x))
print(softmax(x))
print(cross_entropy(softmax(x), x))
```
