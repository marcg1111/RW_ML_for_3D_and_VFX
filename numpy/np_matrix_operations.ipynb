```python
import numpy as np

a = np.array([[3,3,4],
            [4,2,9],
            [9,3,7]])
a
```

```python
b = a.T
b
```

```python
c = np.copy(a)
c
```

```python
c[0, 2] = 10
c
```

```python
c[1, 1] = 25
c
```

```python
e = b.dot(c)
e
```

```python
b @ c
```

```python
f = np.where(e < 100, e, 10*e)
f
```
